```python
import sys

input = sys.stdin.readline
bingo = [list(map(int, input().split())) for _ in range(5)]
noti = []
for _ in range(5):
    noti+= list(map(int, input().split())) 

checking = [[0,0,0,0,0] for _ in range(5)]

def bingotf(x):
    bingocnt = 0
    # 가로 빙고 체크 
    for i in range(5):
        if sum(list(zip(*x))[i]) == 5:
            bingocnt+=1
    # 세로 빙고 체크 
    for i in range(5):
        if sum(x[i]) == 5:
            bingocnt+=1
    # 대각선 빙고 체크 
    if sum([x[i][i] for i in range(0,5)]) == 5:
        bingocnt+=1
    if sum([x[i][4-i] for i in range(0,5)]) == 5:
        bingocnt+=1

    return bingocnt 

cnt=0
for idx, num in enumerate(noti):
    cnt+=1
    # 빙고판 체킹
    for i in range(5):
        for j in range(5):
            if bingo[i][j] == num:
                checking[i][j] = 1

    if idx>4: # 빙고는 최소 5번부터 가능하므로 
        if bingotf(checking) >= 3:
            print(cnt)
            break
```
