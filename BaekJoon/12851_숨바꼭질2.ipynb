```python
from collections import deque
import sys

input=sys.stdin.readline

n,k=map(int,input().split())
visited=[-1 for _ in range(100001)]
cnt=0

def bfs(x,g):
    q=deque()
    q.append(x)
    visited[x]=0
    global cnt

    while q:
        v=q.popleft()
        if v == g:
            cnt+=1

        for i in (2*v,v+1,v-1):
            if 0<=i<=100000 and (visited[i]==-1 or visited[i]>=visited[v]+1): 
                visited[i]=visited[v]+1 # case 저장 
                q.append(i)
                
bfs(n,k)
print(visited[k])
print(cnt)
```
