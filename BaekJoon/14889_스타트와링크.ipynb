```python
import sys
from itertools import combinations

# input = sys.stdin.readline
N = int(input())
lst=[]
for _ in range(N):
    lst.append(list(map(int, input().split())))

min_ans = 100 * 20

for combi in combinations([i for i in range(N)], N//2):
    s1 = s2 = 0
    for i in combi:
        for j in combi:
            s1 += lst[i][j]
            
    score = set(range(N)) - set(combi)
    for i in score:
        for j in score:
            s2+=lst[i][j]
    min_ans=min(min_ans,abs(s1-s2))
print(min_ans)
```
