```python
import sys

input = sys.stdin.readline
total = int(input())
prices = list(map(int, input().split()))

# 준현
cnt_jh = 0
total_jh = total

for price in prices:
    cnt_jh += (total_jh//price)
    total_jh = total_jh%price
    if total_jh == 0 :
        break

# 성민 
cnt_sm = 0
total_sm = total

for i in range(3,len(prices)):
    if prices[i-3] < prices[i-2] < prices[i-1] < prices[i]: # 전량 매도 
        total_sm += (cnt_sm*prices[i])
        cnt_sm = 0

    elif prices[i-3] > prices[i-2] > prices[i-1] > prices[i]: # 전량 매수 
        cnt_sm += total_sm//prices[i]
        total_sm = total_sm%prices[i]

    if (cnt_sm == 0) and (total_sm == 0):
        break

### ans 
ans_jh = total_jh + prices[-1]*cnt_jh
ans_sm = total_sm + prices[-1]*cnt_sm

if ans_jh > ans_sm :
    print('BNP')
elif ans_jh < ans_sm :
    print('TIMING')
else:
    print('SAMESAME')
```
