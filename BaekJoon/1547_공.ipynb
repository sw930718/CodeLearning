```python
import sys

input = sys.stdin.readline
M = int(input())
ball_loc = [1,0,0]

for _ in range(M):
    num1, num2 = map(int, input().split())
    now = ball_loc.index(1)
    if (now+1) in [num1, num2]:
        tmp = ball_loc[num1-1]
        ball_loc[num1-1] = ball_loc[num2-1]
        ball_loc[num2-1] = tmp 
    else:
        continue

print(ball_loc.index(1)+1)
```
