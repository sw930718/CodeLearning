```python
import sys

input = sys.stdin.readline
N = int(input())
counseling = [list(map(int, input().split())) for _ in range(N)]

reservation = [0 for i in range(N+1)]

for i in range(N):
    for j in range(i+counseling[i][0], N+1):
        if reservation[j] < reservation[i] + counseling[i][1]:
            reservation[j] = reservation[i] + counseling[i][1]

print(reservation[-1])
        
```
